```python
import subprocess
from pathlib import Path

repo = Path.cwd().parent
for target in ("patch", "build"):
    r = subprocess.run(["make", target], cwd=repo, capture_output=True, text=True)
    print(f"make {target}: exit {r.returncode}")
    assert r.returncode == 0, r.stdout[-2000:] + r.stderr[-2000:]

r = subprocess.run(["make", "test"], cwd=repo, capture_output=True, text=True)
print(r.stdout)
run_dir = sorted((repo / "logs/p0").iterdir())[-1]
print((run_dir / "SUMMARY.md").read_text())
print((run_dir / "dpf-equivalence.log").read_text().splitlines()[-5:])
```
